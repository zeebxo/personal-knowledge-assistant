```python
from pypdf import PdfReader

reader = PdfReader("Ecommerce_Starter_Guide.pdf")

document_text = ""

for page in reader.pages:
    document_text += page.extract_text() + "\n"

print(document_text[:1000])
```

```python
print("Characters extracted:", len(document_text))
print("Pages extracted:", len(reader.pages))
```

```python
# Step 3: Split document into smaller chunks

chunk_size = 1000
overlap = 200

chunks = []

start = 0

while start < len(document_text):
    end = start + chunk_size
    chunks.append(document_text[start:end])
    start += chunk_size - overlap

print("Number of chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
# Step 3: Split document into overlapping chunks

def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []

    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap

    return chunks


chunks = chunk_text(document_text)

print(f"Created {len(chunks)} chunks")
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = embedder.encode(chunks)

print("Embeddings created successfully!")
print("Number of embeddings:", len(chunk_embeddings))
print("Embedding size:", len(chunk_embeddings[0]))
```

```python
# Step 4: Test document retrieval

import numpy as np

def retrieve(query, top_k=3):
    query_embedding = embedder.encode([query])[0]

    similarities = np.dot(chunk_embeddings, query_embedding) / (
        np.linalg.norm(chunk_embeddings, axis=1) *
        np.linalg.norm(query_embedding)
    )

    top_indices = np.argsort(similarities)[-top_k:][::-1]

    return [(chunks[i], similarities[i]) for i in top_indices]


# Test question
query = "What is e-commerce?"

results = retrieve(query)

for i, (chunk, score) in enumerate(results, 1):
    print(f"\n--- Result {i} | Similarity: {score:.4f} ---\n")
    print(chunk)
```

# Personal Knowledge Assistant

A multi-agent Personal Knowledge Assistant built with CrewAI. The system uses a Researcher agent to search uploaded documents, a Verifier agent to check information on the web, and a Writer agent to generate a clear final answer.

## How to Run

1. Open the project notebook in Google Colab.
2. Install the required Python packages.
3. Add the required API keys to Colab Secrets.
4. Upload the required PDF document.
5. Run the notebook cells from top to bottom.
6. Ask questions through the CrewAI agents.

## Sample Q&A

### Question 1
**What is e-commerce?**

**Answer:** E-commerce is the buying and selling of goods and services over the internet.

### Question 2
**What are some benefits of e-commerce?**

**Answer:** E-commerce can provide convenience, wider market reach, and the ability to operate online.

### Question 3
**What information is available in the uploaded document?**

**Answer:** The Researcher agent searches the uploaded document for relevant information, while the Verifier checks related information using web search.
