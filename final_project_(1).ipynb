```python
from pypdf import PdfReader

reader = PdfReader("Ecommerce_Starter_Guide.pdf")

document_text = ""

for page in reader.pages:
    document_text += page.extract_text() + "\n"

print(document_text[:1000])
```

```python
print("Characters extracted:", len(document_text))
print("Pages extracted:", len(reader.pages))
```

```python
# Step 3: Split document into smaller chunks

chunk_size = 1000
overlap = 200

chunks = []

start = 0

while start < len(document_text):
    end = start + chunk_size
    chunks.append(document_text[start:end])
    start += chunk_size - overlap

print("Number of chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
# Step 3: Split document into overlapping chunks

def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []

    start = 0

    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap

    return chunks


chunks = chunk_text(document_text)

print(f"Created {len(chunks)} chunks")
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

chunk_embeddings = embedder.encode(chunks)

print("Embeddings created successfully!")
print("Number of embeddings:", len(chunk_embeddings))
print("Embedding size:", len(chunk_embeddings[0]))
```

```python
# Step 4: Test document retrieval

import numpy as np

def retrieve(query, top_k=3):
    query_embedding = embedder.encode([query])[0]

    similarities = np.dot(chunk_embeddings, query_embedding) / (
        np.linalg.norm(chunk_embeddings, axis=1) *
        np.linalg.norm(query_embedding)
    )

    top_indices = np.argsort(similarities)[-top_k:][::-1]

    return [(chunks[i], similarities[i]) for i in top_indices]


# Test question
query = "What is e-commerce?"

results = retrieve(query)

for i, (chunk, score) in enumerate(results, 1):
    print(f"\n--- Result {i} | Similarity: {score:.4f} ---\n")
    print(chunk)
```

# step 5 - 6

```python
from crewai.tools import tool

@tool("Document Search")
def document_search(query: str) -> str:
    """Searches the student's own uploaded documents for relevant passages."""
    results = retrieve(query)
    return str(results)
```

```python
researcher = Agent(
    role="Document Researcher",
    goal="Search the student's uploaded documents and find passages relevant to the user's question.",
    backstory="You are a careful research assistant who answers questions by searching the student's own uploaded documents.",
    tools=[document_search],
    llm=llm
)
```

```python
from crewai.tools import tool
from ddgs import DDGS

@tool("Web Search")
def web_search(query: str) -> str:
    """Searches the live web for information related to the user's question."""
    results = DDGS().text(query, max_results=5)
    return str(results)
```

```python
verifier = Agent(
    role="Web Verifier",
    goal="Verify the Researcher's findings by checking relevant information on the live web.",
    backstory="You are a careful fact-checker who uses web search to verify information found in the student's documents.",
    tools=[web_search],
    llm=llm
)
```

```python
research_task = Task(
    description="Answer the question using information from the student's uploaded documents. Question: What is e-commerce?",
    expected_output="A clear answer based on relevant passages from the uploaded documents.",
    agent=researcher
)

verify_task = Task(
    description="Verify the Researcher's findings by searching the live web for relevant information about e-commerce.",
    expected_output="A brief verification of the Researcher's findings using web sources.",
    agent=verifier
)
```

```python
writer = Agent(
    role="Answer Writer",
    goal="Synthesize the Researcher's findings and the Verifier's web verification into one clear and accurate answer.",
    backstory="You are a careful writer who combines research findings and verification results into a concise final response.",
    llm=llm
)
```

```python
research_task = Task(
    description="Answer the user's question: {question}. Search the student's uploaded documents and find relevant information.",
    agent=researcher,
    expected_output="Relevant findings from the uploaded documents that answer the user's question."
)

verify_task = Task(
    description="Verify the Researcher's findings for the user's question: {question}. Use web search to check whether the findings are supported.",
    agent=verifier,
    context=[research_task],
    expected_output="A verification of the Researcher's findings using web sources."
)

write_task = Task(
    description="Combine the Researcher's findings and the Verifier's verification into one clear, accurate final answer.",
    agent=writer,
    context=[verify_task],
    expected_output="A clear and concise final answer to the user's question."
)

crew = Crew(
    agents=[researcher, verifier, writer],
    tasks=[research_task, verify_task, write_task],
    process=Process.sequential
)
```
